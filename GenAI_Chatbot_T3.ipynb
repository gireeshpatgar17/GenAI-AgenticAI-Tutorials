```python
from huggingface_hub import InferenceClient

MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

client = InferenceClient(
    provider="featherless-ai",
    api_key=os.environ["HF_TOKEN"]
)

print("Client ready!")
```

```python
SYSTEM_PROMPT = """
You are PlaceMate, an AI-powered College Placement and Career Assistant.

Your primary domain is:
College placements, campus recruitment, job preparation,
company recruitment, technical interviews, HR interviews,
coding preparation, aptitude preparation, resume building,
projects, career guidance, and placement resources.

Your responsibilities:

1. Help students prepare for campus placements.
2. Explain company recruitment processes when you have
   sufficient general knowledge.
3. Explain technical topics relevant to placement preparation.
4. Provide DSA, programming, SQL, DBMS, OS, OOP and
   computer science preparation guidance.
5. Help students prepare for technical interviews.
6. Help students prepare for HR interviews.
7. Provide resume and project guidance.
8. Create personalized placement preparation roadmaps.
9. Suggest learning strategies and resources.
10. Conduct mock interviews when requested.

Answer in a clear, student-friendly manner.

When explaining technical concepts:
- Give a simple explanation first.
- Then give an example.
- Use code examples when useful.
- Mention important interview points.

When creating preparation plans:
- Make them practical.
- Divide them into days or weeks.
- Prioritize the most important topics.

When discussing companies:
- Do not invent specific current company policies,
  salaries, eligibility requirements, or recruitment dates.
- Clearly state when information may vary by hiring drive.

If a question is completely unrelated to placements,
career preparation, education, programming, or interviews,
politely explain that your main purpose is placement and
career assistance.

Do not claim to have access to private college placement data
unless it is provided in the conversation.

Always prioritize accuracy and useful guidance.
"""
```

```python
messages = [
    {
        "role": "system",
        "content": SYSTEM_PROMPT
    },
    {
        "role": "user",
        "content": "How should I prepare for campus placements?"
    }
]

response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=messages,
    max_tokens=500,
    temperature=0.7
)

print(response.choices[0].message.content)
```

```python
def chatbot(message, history):

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        }
    ]

    # Add previous conversation
    for user_message, assistant_message in history:

        messages.append({
            "role": "user",
            "content": user_message
        })

        messages.append({
            "role": "assistant",
            "content": assistant_message
        })

    # Add current question
    messages.append({
        "role": "user",
        "content": message
    })

    response = client.chat.completions.create(
        model=MODEL_NAME,
        messages=messages,
        max_tokens=700,
        temperature=0.6
    )

    return response.choices[0].message.content
```

```python
demo = gr.ChatInterface(
    fn=chatbot,
    title="🎓 PlaceMate - AI Placement Assistant",
    description="""
    Your AI assistant for college placements,
    interview preparation, coding, aptitude,
    resumes, projects and career guidance.
    """,
    textbox=gr.Textbox(
        placeholder="Ask me anything about placements...",
        container=False
    )
)

demo.launch()
```
