```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model ="gpt2"
)
```

```python
output = generator(
    "Artificial intelligence is",
    max_new_tokens=30
)
print(output)
```

```python
output = generator(
    "The future of Artificial intelligence is",
    max_new_tokens=50
)
print(output)
```

```python
output = generator(
    "The future of Artificial intelligence is",
    max_new_tokens=50,
    temperature=0.2,
    do_sample=True
)
print(output)
```

```python
output = generator(
    "The future of Artificial intelligence is",
    max_new_tokens=50,
    temperature=1.0,
    do_sample=True
)
print(output)
```

```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```python
text = "Generative AI is growing exponentially in these days."
```

```python
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
token_ids = tokenizer.encode(text)
print(token_ids)
```

```python
for token, token_id in zip(tokens, token_ids):
  print(token, "->", token_id)
```
